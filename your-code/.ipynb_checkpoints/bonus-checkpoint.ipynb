# Challenge 1 - T-test

In statistics, t-test is used to test if two data samples have a significant difference between their means. There are two types of t-test:

* **Student's t-test** (a.k.a. independent or uncorrelated t-test). This type of t-test is to compare the samples of **two independent populations** (e.g. test scores of students in two different classes). `scipy` provides the [`ttest_ind`](https://docs.scipy.org/doc/scipy-0.15.1/reference/generated/scipy.stats.ttest_ind.html) method to conduct student's t-test.

* **Paired t-test** (a.k.a. dependent or correlated t-test). This type of t-test is to compare the samples of **the same population** (e.g. scores of different tests of students in the same class). `scipy` provides the [`ttest_re`](https://docs.scipy.org/doc/scipy-0.15.1/reference/generated/scipy.stats.ttest_rel.html) method to conduct paired t-test.

Both types of t-tests return a number which is called the **p-value**. If p-value is below 0.05, we can confidently declare the null-hypothesis is rejected and the difference is significant. If p-value is between 0.05 and 0.1, we may also declare the null-hypothesis is rejected but we are not highly confident. If p-value is above 0.1 we do not reject the null-hypothesis.

Read more about the t-test in [this article](https://researchbasics.education.uconn.edu/t-test/) and [this Quora](https://www.quora.com/What-is-the-difference-between-a-paired-and-unpaired-t-test). Make sure you understand when to use which type of t-test.

## Independent t‑test Completed Function

````python
from scipy.stats import ttest_ind

def t_test_features(s1, s2, features=['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed', 'Total']):
    """Test means of a feature set of two samples
    
    Args:
        s1 (dataframe): sample 1
        s2 (dataframe): sample 2
        features (list): an array of features to test
    
    Returns:
        dict: a dictionary of t-test scores for each feature where the feature name is the key and the p-value is the value
    """
    results = {}

    for feature in features:
        # extract the feature values from each sample
        x1 = s1[feature]
        x2 = s2[feature]

        # perform independent t-test (Welch’s t-test recommended)
        stat, p = ttest_ind(x1, x2, equal_var=False)

        # store p-value in dictionary
        results[feature] = p

    return results
````

## Paired t‑test Function (Feature‑to‑Feature Within Same Sample)

````python
from scipy.stats import ttest_rel

def paired_t_test_features(df, feature_pairs = [("Attack", "Sp. Atk"), ("Defense", "Sp. Def"), ("HP", "Speed")]):
    """
    Perform paired t-tests on pairs of features from the same dataframe.
    
    Args:
        df (dataframe): dataset containing the features
        feature_pairs (list of tuples): list of (feature1, feature2) pairs to compare
    
    Returns:
        dict: dictionary where keys are "feature1 vs feature2" 
              and values are the p-values of the paired t-test
    """
    results = {}

    for f1, f2 in feature_pairs:
        x1 = df[f1]
        x2 = df[f2]

        stat, p = ttest_rel(x1, x2)
        results[f"{f1} vs {f2}"] = p

    return results
````

````python
# Split dataset into Legendary and non-Legendary groups
legendary = df[df["Legendary"] == True]
non_legendary = df[df["Legendary"] == False]

# Run the t-tests
results = t_test_features(legendary, non_legendary)

# Display results
results
````
