# Challenge 2 - Hypothesis Tests

In this section of the lab, we will test whether the hourly wage of all hourly workers is significantly different from $30/hr. Import the correct one sample test function from scipy and perform the hypothesis test for a 95% two sided confidence interval.

## Goal of the Hypothesis Test

We test:

$$ H_0: \mu = 30 $$ $$ H_a: \mu \neq 30 $$

This is a **two‑sided one‑sample t‑test** because:
* We have **one sample** (hourly workers)
* We compare the sample mean to a **known value** (30)
* Population σ is **unknown** → we use **t‑test**

## Step‑by‑Step Code (with heavy comments)

```python
from scipy.stats import ttest_1samp   # correct one-sample t-test

# ---------------------------------
# - 1. Filter hourly workers only -
# ---------------------------------
hourly_salaries = salaries[salaries["Salary or Hourly"] == "Hourly"]

# Extract the hourly rate column as a numeric array
hourly_rates = pd.to_numeric(hourly_salaries["Hourly Rate"], errors="coerce").dropna()

# -----------------------------------
# - 2. Define the hypothesized mean -
# -----------------------------------
mu0 = 30   # $30/hr

# ------------------------------------
# - 3. Perform the one-sample t-test -
# ------------------------------------
t_stat, p_value = ttest_1samp(hourly_rates, popmean=mu0)

print("T-statistic:", t_stat)
print("P-value:", p_value)

# ------------------------------------------
# - 4. Compute the 95% confidence interval -
# ------------------------------------------
import scipy.stats as stats

n = len(hourly_rates)
xbar = hourly_rates.mean()
s = hourly_rates.std(ddof=1)
alpha = 0.05

# Critical t-value for 95% CI (two-sided)
t_crit = stats.t.ppf(1 - alpha/2, df=n-1)

# Margin of error
margin = t_crit * (s / np.sqrt(n))

ci_lower = xbar - margin
ci_upper = xbar + margin

print("95% Confidence Interval:", (ci_lower, ci_upper))
```

## How to Interpret the Results

1. ### Decision Rule

* If **p-value < 0.05**, reject (H_0) → hourly wage is significantly different from $30
* If **p-value ≥ 0.05**, fail to reject (H_0) → no evidence of difference

2. ### Confidence Interval Interpretation

* If **30 is outside the CI**, reject (H_0)
* If **30 is inside the CI**, fail to reject (H_0)

Both the p‑value and CI must agree.

# Challenge 3 - Constructing Confidence Intervals

While testing our hypothesis is a great way to gather empirical evidence for accepting or rejecting the hypothesis, another way to gather evidence is by creating a confidence interval. A confidence interval gives us information about the true mean of the population. So for a 95% confidence interval, we are 95% sure that the mean of the population is within the confidence interval. 
).

To read more about confidence intervals, click [here](https://en.wikipedia.org/wiki/Confidence_interval).


In the cell below, we will construct a 95% confidence interval for the mean hourly wage of all hourly workers. 

The confidence interval is computed in SciPy using the `t.interval` function. You can read more about this function [here](https://docs.scipy.org/doc/scipy-0.14.0/reference/generated/scipy.stats.t.html).

To compute the confidence interval of the hourly wage, use the 0.95 for the confidence level, number of rows - 1 for degrees of freedom, the mean of the sample for the location parameter and the standard error for the scale. The standard error can be computed using [this](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.sem.html) function in SciPy.

## 95% Confidence Interval for Hourly Wages

We already computed:

* **n = 8022**
* **mean = 32.788558**
* **std = 12.112573**

The standard error is:

$$ SE = \frac{12.112573}{\sqrt{8022}} \approx 0.1352 $$

The 95% CI is:

$$ 32.7886 \pm 1.96 \cdot 0.1352 $$

$$ CI = (32.523,\ 33.054) $$

```python


# Extract hourly workers
hourly_df = salaries[salaries["Salary or Hourly"] == "Hourly"]
hourly_rates = pd.to_numeric(hourly_df["Hourly Rate"], errors="coerce").dropna()

# Compute statistics
mean_hourly = hourly_rates.mean()
sem_hourly = stats.sem(hourly_rates)
dfree = len(hourly_rates) - 1

# 95% confidence interval
ci_low, ci_high = stats.t.interval(
    confidence=0.95,
    df=dfree,
    loc=mean_hourly,
    scale=sem_hourly
)

(ci_low, ci_high)
```

### Interpretation

We are **95% confident** that the true mean hourly wage of hourly workers in Chicago lies between: **32.52 and $33.05**

Since **30 is not inside this interval**, the data strongly suggests that the true mean hourly wage is **different from $30/hr**.

This matches the conclusion from our earlier **one‑sample t‑test**.



# Challenge 4 - Hypothesis Tests of Proportions

Another type of one sample test is a hypothesis test of proportions. In this test, we examine whether the proportion of a group in our sample is significantly different than a fraction. 

You can read more about one sample proportion tests [here](http://sphweb.bumc.bu.edu/otlt/MPH-Modules/BS/SAS/SAS6-CategoricalData/SAS6-CategoricalData2.html).

In the cell below, use the `proportions_ztest` function from `statsmodels` to perform a hypothesis test that will determine whether the number of hourly workers in the City of Chicago is significantly different from 25% at the 95% confidence level.

## What we are testing

Let:

* $ p $ = true proportion of hourly workers in Chicago
* Hypothesized proportion = **0.25**

### Hypotheses

$$ H_0: p = 0.25 $$  $$ H_a: p \neq 0.25 $$

This is a **two‑sided one‑sample proportion test**.

## Step‑by‑Step Code (fully commented)

```python
from statsmodels.stats.proportion import proportions_ztest

# ---------------------------
# - 1. Count hourly workers -
# ---------------------------
hourly_count = (salaries["Salary or Hourly"] == "Hourly").sum()

# 2. Total number of employees
total_count = len(salaries)

# 3. Hypothesized proportion
p0 = 0.25   # 25%

# ------------------------------------------------
# - 4. Perform one-sample z-test for proportions -
# ------------------------------------------------
z_stat, p_value = proportions_ztest(
    count=hourly_count,     # number of hourly workers
    nobs=total_count,       # total employees
    value=p0                # hypothesized proportion
)

print("Hourly workers:", hourly_count)
print("Total employees:", total_count)
print("Sample proportion:", hourly_count / total_count)
print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

## Interpretation

We interpret the result like this:

### Decision rule

* If **p-value < 0.05**, reject (H_0) → the proportion is significantly different from 25%.
* If **p-value ≥ 0.05**, fail to reject (H_0) → no evidence the proportion differs from 25%.

### What the test tells us

This test answers:

“Is the percentage of hourly workers in Chicago different from 25%?”
